```python
import pandas as pd

# Load the datasets with the correct delimiter
train_df = pd.read_csv('train_dataset.csv', delimiter=';')
test_df = pd.read_csv('test_dataset.csv', delimiter=';')

# Display the first few rows and column names to verify
print(train_df.head())
print(train_df.columns)
```

```python
# Assuming 'class' is the target column after correcting the delimiter
target_column = 'class'  # Update if needed

X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]
X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]
```

```python
from sklearn.preprocessing import LabelEncoder

# Prepare features and target variable
X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]
X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]

# Encode target labels if they are categorical
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_columns = X_train.select_dtypes(include=['object']).columns

# One-hot encode categorical features
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), categorical_columns)
    ],
    remainder='passthrough'
)

X_train_preprocessed = preprocessor.fit_transform(X_train)
X_test_preprocessed = preprocessor.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Train the Logistic Regression model
logreg = LogisticRegression(max_iter=1000)
logreg.fit(X_train_preprocessed, y_train_encoded)

# Make predictions
logreg_predictions = logreg.predict(X_test_preprocessed)

# Evaluate the model
print("Logistic Regression Accuracy:", accuracy_score(y_test_encoded, logreg_predictions))
print(classification_report(y_test_encoded, logreg_predictions))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Train the Random Forest model
rf = RandomForestClassifier()
rf.fit(X_train_preprocessed, y_train_encoded)

# Make predictions
rf_predictions = rf.predict(X_test_preprocessed)

# Evaluate the model
print("Random Forest Accuracy:", accuracy_score(y_test_encoded, rf_predictions))
print(classification_report(y_test_encoded, rf_predictions))
```

```python
# Choose the best model based on accuracy or any other metric
# For demonstration, we'll save predictions from both models

# Logistic Regression predictions
logreg_results = pd.DataFrame({
    'True Label': y_test,
    'Logistic Regression Prediction': label_encoder.inverse_transform(logreg_predictions)
})

# Random Forest predictions
rf_results = pd.DataFrame({
    'True Label': y_test,
    'Random Forest Prediction': label_encoder.inverse_transform(rf_predictions)
})

# Save results to CSV
logreg_results.to_csv('logreg_predictions.csv', index=False)
rf_results.to_csv('rf_predictions.csv', index=False)

# Download the CSV files
from google.colab import files
files.download('logreg_predictions.csv')
files.download('rf_predictions.csv')
```
